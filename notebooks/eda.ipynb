```python
fig, axes = plt.subplots(3, 1, figsize=(10, 9))

hard_faults = [3, 9, 15]

for i, f in enumerate(hard_faults):
    fault_data = test[test["faultNumber"] == f]
    axes[i].plot(normal["XMEAS_9"].values, label="Normal", alpha=0.7)
    axes[i].plot(fault_data["XMEAS_9"].values, label=f"Fault {f}", alpha=0.7)
    axes[i].set_title(f"Fault {f}")
    axes[i].legend()

plt.tight_layout()
plt.show()
```
