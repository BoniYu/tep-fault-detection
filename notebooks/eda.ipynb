```python
import seaborn as sns

normal_train = train[train["faultNumber"] == 0]
measurement_cols = [c for c in train.columns if c.startswith("XMEAS") or c.startswith("XMV")]

corr = normal_train[measurement_cols].corr()

plt.figure(figsize=(14, 12))
sns.heatmap(corr, cmap="coolwarm", center=0)
plt.title("Variable correlation — normal operation")
plt.tight_layout()
plt.show()
```

```python
"""Pulls the top strongest variable correlations (positive or negative)
from the heatmap above, printed with readable names instead of column
codes, so we can see exactly which pairs are driving those bright
red/blue spots."""

corr_pairs = corr.abs().unstack()
corr_pairs = corr_pairs[corr_pairs < 1.0]  # drop self-correlation (always 1.0)
corr_pairs = corr_pairs.sort_values(ascending=False)

#top_pairs = corr_pairs.drop_duplicates().head(10) #each pair actually shows up twice (A-B and B-A) with the same value, keep only one of them.
#for (var1, var2), value in top_pairs.items():
    #print(f"{value:.3f}  {describe_column(var1)}  <->  {describe_column(var2)}")

threshold = 0.85

strong_pairs = corr_pairs[corr_pairs > threshold].drop_duplicates()

print(f"{len(strong_pairs)} pairs above {threshold} correlation:\n")
for (var1, var2), value in strong_pairs.items():
    print(f"{value:.3f}  {describe_column(var1)}  <->  {describe_column(var2)}")    
```

```python
# PCA 
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler

faults_to_check = [0, 1, 4, 6, 13, 14]  # a mix of faults to keep the plot readable
subset = train[train["faultNumber"].isin(faults_to_check)]

X = StandardScaler().fit_transform(subset[measurement_cols])

pca = PCA(n_components=2)
X_pca = pca.fit_transform(X)

plt.figure(figsize=(9, 7))
for f in faults_to_check:
    mask = subset["faultNumber"].values == f
    plt.scatter(X_pca[mask, 0], X_pca[mask, 1], label=describe(f), alpha=0.4, s=10)

plt.xlabel(f"PC1 ({pca.explained_variance_ratio_[0]:.1%} of variance)")
plt.ylabel(f"PC2 ({pca.explained_variance_ratio_[1]:.1%} of variance)")
plt.title("PCA — 2D projection of selected faults (train data)")
plt.legend(fontsize=8)
plt.tight_layout()
plt.show()

```

```python
pca_full = PCA()
pca_full.fit(StandardScaler().fit_transform(train[measurement_cols]))

variance_per_component = pca_full.explained_variance_ratio_
cumulative_variance = variance_per_component.cumsum()

fig, ax1 = plt.subplots(figsize=(12, 5))

ax1.bar(range(1, len(variance_per_component) + 1), variance_per_component, alpha=0.6, label="Per component")
ax1.set_xlabel("Number of components")
ax1.set_ylabel("Variance explained (this component)")

ax2 = ax1.twinx()
ax2.plot(range(1, len(cumulative_variance) + 1), cumulative_variance, color="red", marker="o", markersize=3, label="Cumulative")
ax2.set_ylabel("Cumulative variance explained")
ax2.axhline(0.90, color="gray", linestyle="--", alpha=0.5)

plt.title("PCA — variance explained by number of components")
fig.tight_layout()
plt.show()

# Find how many components hit common thresholds
for threshold in [0.80, 0.90, 0.95]:
    n_components = (cumulative_variance >= threshold).argmax() + 1
    print(f"{threshold:.0%} variance explained by {n_components} components")
```

```python
#Checking from where the fault starts
import numpy as np

for f in range(1, 22):
    fault_data = test[test["faultNumber"] == f].reset_index(drop=True)
    
    # Compare each sample's deviation from the normal baseline
    deviation = (fault_data[measurement_cols] - normal[measurement_cols].mean()).abs().sum(axis=1)
    
    # Rolling average smooths out noise so we can spot the real shift point
    smoothed = deviation.rolling(10, min_periods=1).mean()
    
    baseline = smoothed[:150].mean()
    threshold = baseline * 1.5
    
    above_threshold = smoothed[smoothed > threshold]
    onset = above_threshold.index[0] if len(above_threshold) > 0 else None
    
    print(f"Fault {f}: detected onset around sample {onset}")
```

```python
# SPlitting into X and y for model training and evaluation

X_train_all = train_all[measurement_cols]
y_train_all = train_all["faultNumber"]

X_train_filtered = train_filtered[measurement_cols]
y_train_filtered = train_filtered["faultNumber"]

X_test_all = test_all[measurement_cols]
y_test_all = test_all["faultNumber"]

X_test_filtered = test_filtered[measurement_cols]
y_test_filtered = test_filtered["faultNumber"]

print("X_train_all:", X_train_all.shape, " y_train_all:", y_train_all.shape)
print("X_train_filtered:", X_train_filtered.shape, " y_train_filtered:", y_train_filtered.shape)
print("X_test_all:", X_test_all.shape, " y_test_all:", y_test_all.shape)
print("X_test_filtered:", X_test_filtered.shape, " y_test_filtered:", y_test_filtered.shape)
```

**Findings**: Top variables are dominated by control valves paired
with the measurements they directly regulate — XMV_4 (A/C feed valve),
XMV_10 (reactor cooling water flow), XMEAS_21 (cooling water outlet
temp), XMEAS_1 (A feed rate), and the stripper steam trio (XMV_9,

XMV_10 ranking #2 overall directly confirms the earlier EDA finding
that Fault 4's signature lives in the compensating cooling water valve,
not reactor temperature (XMEAS_9) itself — now validated by the
model's learned importances across all 19 faults, not just Fault 4.

Overall, the model's reasoning aligns with real control relationships
identified in EDA, rather than obscure or uninterpretable patterns —
a good sign predictions are grounded in genuine process behavior.

```python
# XGBoost Classifier for the filtered dataset
from xgboost import XGBClassifier
from sklearn.preprocessing import LabelEncoder

label_encoder = LabelEncoder()
y_train_filtered_encoded = label_encoder.fit_transform(y_train_filtered) #3,9,15 are excluded from the filtered dataset
y_test_filtered_encoded = label_encoder.transform(y_test_filtered)

xgb_filtered = XGBClassifier(
    n_estimators=100,
    random_state=42,
    n_jobs=-1,
    eval_metric="mlogloss"
)

start = time.time()
xgb_filtered.fit(X_train_filtered, y_train_filtered_encoded)
elapsed = time.time() - start

print(f"Trained in {elapsed:.1f} seconds")
```

```python
y_pred_filtered_encoded = xgb_filtered.predict(X_test_filtered)

y_pred_xgb = label_encoder.inverse_transform(y_pred_filtered_encoded)

accuracy_xgb = accuracy_score(y_test_filtered, y_pred_xgb)
print(f"XGBoost accuracy (excluding 3/9/15): {accuracy_xgb:.2%}")
print(f"Random Forest accuracy (excluding 3/9/15): {accuracy_filtered:.2%}\n")

print(classification_report(y_test_filtered, y_pred_xgb))
```

```python
cm_xgb = confusion_matrix(y_test_filtered, y_pred_xgb)
fault_labels_xgb = sorted(y_test_filtered.unique())

for fault_num in [10, 16]:
    idx = fault_labels_xgb.index(fault_num)
    row = cm_xgb[idx]
    print(f"\nTrue fault {fault_num} — model predicted:")
    for pred_num, count in zip(fault_labels_xgb, row):
        if count > 20:
            print(f"  Fault {pred_num}: {count}")
```

```python
# Hyperparameter tuning for XGBoost using RandomizedSearchCV
from sklearn.model_selection import RandomizedSearchCV

param_distributions = {
    "n_estimators": [50, 100, 200, 300],
    "max_depth": [3, 5, 7, 10],
    "learning_rate": [0.01, 0.05, 0.1, 0.2],
    "subsample": [0.7, 0.8, 1.0],
}

xgb_search = RandomizedSearchCV(
    XGBClassifier(random_state=42, n_jobs=-1, eval_metric="mlogloss"),
    param_distributions=param_distributions,
    n_iter=15,
    cv=3,
    random_state=42,
    verbose=2,
)

start = time.time()
xgb_search.fit(X_train_filtered, y_train_filtered_encoded)
elapsed = time.time() - start

print(f"\nSearch completed in {elapsed/60:.1f} minutes")
print(f"Best parameters: {xgb_search.best_params_}")
print(f"Best cross-validation accuracy: {xgb_search.best_score_:.2%}")
```

```python
# Training with the best parameters found from RandomizedSearchCV

xgb_tuned = XGBClassifier(
    n_estimators=200,
    max_depth=7,
    learning_rate=0.05,
    subsample=0.7,
    random_state=42,
    n_jobs=-1,
    eval_metric="mlogloss"
)

xgb_tuned.fit(X_train_filtered, y_train_filtered_encoded)

y_pred_tuned_encoded = xgb_tuned.predict(X_test_filtered)
y_pred_tuned = label_encoder.inverse_transform(y_pred_tuned_encoded)

accuracy_tuned = accuracy_score(y_test_filtered, y_pred_tuned)

print(f"XGBoost (default settings): {accuracy_xgb:.2%}")
print(f"XGBoost (tuned):            {accuracy_tuned:.2%}")
```

```python
print(classification_report(y_test_filtered, y_pred_tuned))
```

## Trying rolling/rate-of-change features

Faults 10, 13, and 16 have remained the weakest performers across
every approach so far (filtering, XGBoost, tuning) — modeling so far
only sees single-row snapshots, with no sense of whether a variable is
rising, falling, or steady. Testing whether adding trend information
(rate-of-change and rolling averages) for the top 15 most important
variables helps the model distinguish these faults, before considering
the more complex LSTM stretch goal.

```python
sys.path.append("../src")
from feature_engineering import add_trend_features

trend_columns = list(top_15.index)  # the 15 most important variables from earlier

train_filtered_trend = add_trend_features(train_filtered, trend_columns)
test_filtered_trend = add_trend_features(test_filtered, trend_columns)

print("New shape:", train_filtered_trend.shape)
print("New columns added:", train_filtered_trend.shape[1] - train_filtered.shape[1])
```

```python
new_measurement_cols = [c for c in train_filtered_trend.columns
                          if c not in ["faultNumber", "sample"]]

X_train_trend = train_filtered_trend[new_measurement_cols]
y_train_trend = train_filtered_trend["faultNumber"]

X_test_trend = test_filtered_trend[new_measurement_cols]
y_test_trend = test_filtered_trend["faultNumber"]

y_train_trend_encoded = label_encoder.transform(y_train_trend)

xgb_trend = XGBClassifier(
    n_estimators=200,
    max_depth=7,
    learning_rate=0.05,
    subsample=0.7,
    random_state=42,
    n_jobs=-1,
    eval_metric="mlogloss"
)

xgb_trend.fit(X_train_trend, y_train_trend_encoded)

y_pred_trend_encoded = xgb_trend.predict(X_test_trend)
y_pred_trend = label_encoder.inverse_transform(y_pred_trend_encoded)

accuracy_trend = accuracy_score(y_test_trend, y_pred_trend)

print(f"XGBoost (tuned, no trend features): {accuracy_tuned:.2%}")
print(f"XGBoost (tuned, with trend features): {accuracy_trend:.2%}")
```

```python
print(classification_report(y_test_trend, y_pred_trend))
```
